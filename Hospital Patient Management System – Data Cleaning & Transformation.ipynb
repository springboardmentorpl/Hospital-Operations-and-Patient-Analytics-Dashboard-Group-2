# IMPORT LIBRARIES

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.width", 120)

print("Libraries imported successfully!")
```

# LOAD DATASET

```python
file_path = "hopa.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])
```

# INITIAL DATA VIEW

```python

display(df.head())
```

# DATASET INFORMATION

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
display(df.dtypes.to_frame("Data Type"))

print("\nDataset Information:")
df.info()
```

# STATISTICAL SUMMARY

```python
display(df.describe(include="all").T)
```

# MISSING VALUE ANALYSIS

```python
missing_report = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (
        df.isnull().mean() * 100
    ).round(2)
})

missing_report = missing_report.sort_values(
    "Missing_Count",
    ascending=False
)

display(missing_report)
```

# MISSING VALUE VISUALIZATION

```python
missing_plot = missing_report[
    missing_report["Missing_Count"] > 0
]

plt.figure(figsize=(14, 6))

plt.bar(
    missing_plot.index,
    missing_plot["Missing_Percentage"]
)

plt.xticks(rotation=90)
plt.ylabel("Missing Percentage (%)")
plt.xlabel("Columns")
plt.title("Missing Values Before Cleaning")

plt.tight_layout()
plt.show()
```

# DUPLICATE CHECK

```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

# STANDARDIZE COLUMN NAMES

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("/", "_")
)

print("Cleaned column names:")
print(df.columns.tolist())
```

# CLEAN TEXT DATA

```python
text_columns = df.select_dtypes(
    include=["object", "string"]
).columns

for col in text_columns:
    df[col] = (
        df[col]
        .astype("string")
        .str.strip()
    )

print("Text columns cleaned successfully!")
```

# CONVERT NUMERIC COLUMNS

```python
numeric_columns = [
    "age",
    "staff_count",
    "capacity",
    "total_bill",
    "daily_rate",
    "length_of_stay_days",
    "length_of_stay_numeric",
    "estimated_bed_cost"
]

for col in numeric_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        )

print("Numeric columns converted successfully!")
```

# CONVERT DATE COLUMNS

```python
date_columns = [
    "admission_date",
    "discharge_date",
    "date_of_birth",
    "registration_date"
]

for col in date_columns:
    if col in df.columns:
        df[col] = pd.to_datetime(
            df[col],
            errors="coerce"
        )

print("Date columns converted successfully!")

display(
    df[date_columns].head()
)
```

#  HANDLE MISSING CATEGORICAL VALUES

```python
categorical_columns = [
    "diagnosis",
    "insurance_provider",
    "insurance_provider_patient",
    "city",
    "state",
    "allergies",
    "chronic_condition",
    "payment_status",
    "discharge_disposition",
    "phone",
    "email"
]

for col in categorical_columns:
    if col in df.columns:
        df[col] = df[col].fillna("Unknown")

print("Categorical missing values handled!")
```

# HANDLE NUMERICAL MISSING VALUES

```python
# Numerical fields where median imputation is reasonable
impute_columns = [
    "age",
    "staff_count",
    "capacity",
    "total_bill",
    "daily_rate"
]

for col in impute_columns:
    if col in df.columns:
        df[col] = df[col].fillna(
            df[col].median()
        )

print("Numerical missing values handled!")
```

# AGE VALIDATION

```python
invalid_age = df[
    (df["age"] < 0) |
    (df["age"] > 120)
]

print("Invalid age records:", len(invalid_age))

if len(invalid_age) > 0:
    print("\nInvalid records:")
    display(
        invalid_age[
            ["patient_id", "date_of_birth", "age"]
        ].head(20)
    )
```

# BILL VALIDATION

```python
invalid_bills = df[
    df["total_bill"] <= 0
]

print("Invalid bill records:", len(invalid_bills))

if len(invalid_bills) > 0:
    display(
        invalid_bills[
            ["admission_id", "patient_id", "total_bill"]
        ].head(20)
    )
```

# LENGTH OF STAY TRANSFORMATION

```python
df["calculated_length_of_stay"] = (
    df["discharge_date"] -
    df["admission_date"]
).dt.days

# Only replace existing LOS when a valid discharge date exists
valid_los = (
    df["calculated_length_of_stay"].notna() &
    (df["calculated_length_of_stay"] >= 0)
)

df.loc[
    valid_los,
    "length_of_stay_numeric"
] = df.loc[
    valid_los,
    "calculated_length_of_stay"
]

print("Length of stay calculated!")
```

#  AGE GROUP

```python
df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 17, 30, 45, 60, 75, 120],
    labels=[
        "0-17",
        "18-30",
        "31-45",
        "46-60",
        "61-75",
        "76+"
    ],
    include_lowest=True
)

display(
    df[
        ["patient_id", "age", "age_group"]
    ].head(10)
)
```

# LENGTH OF STAY CATEGORY

```python
df["stay_category"] = pd.cut(
    df["length_of_stay_numeric"],
    bins=[
        -1,
        2,
        5,
        10,
        np.inf
    ],
    labels=[
        "Short Stay",
        "Medium Stay",
        "Long Stay",
        "Extended Stay"
    ]
)

display(
    df[
        [
            "patient_id",
            "length_of_stay_numeric",
            "stay_category"
        ]
    ].head(10)
)
```

# BILLING CATEGORY

```python
df["billing_category"] = pd.cut(
    df["total_bill"],
    bins=[
        0,
        10000,
        25000,
        50000,
        100000,
        np.inf
    ],
    labels=[
        "Low",
        "Moderate",
        "High",
        "Very High",
        "Critical"
    ]
)

display(
    df[
        [
            "patient_id",
            "total_bill",
            "billing_category"
        ]
    ].head(10)
)
```

# ESTIMATED BED COST

```python
df["estimated_bed_cost"] = (
    df["length_of_stay_numeric"].fillna(0)
    *
    df["daily_rate"].fillna(0)
).round(2)

display(
    df[
        [
            "patient_id",
            "length_of_stay_numeric",
            "daily_rate",
            "estimated_bed_cost"
        ]
    ].head(10)
)
```

# 30-DAY READMISSION ANALYSIS

```python
# Make sure dates are datetime
df["admission_date"] = pd.to_datetime(
    df["admission_date"],
    errors="coerce"
)

# Sort admissions by patient and date
df = df.sort_values(
    ["patient_id", "admission_date"]
).reset_index(drop=True)

# Previous admission date for each patient
df["previous_admission_date"] = (
    df.groupby("patient_id")["admission_date"]
      .shift(1)
)

# Days since previous admission
df["days_since_previous_admission"] = (
    df["admission_date"]
    - df["previous_admission_date"]
).dt.days

# 30-day readmission flag
df["readmission_30_days"] = np.where(
    df["days_since_previous_admission"].between(1, 30),
    "Yes",
    "No"
)

print("30-day readmission analysis completed!")

print("\nReadmission column created:")
print(df["readmission_30_days"].value_counts())

display(
    df[
        [
            "patient_id",
            "admission_date",
            "previous_admission_date",
            "days_since_previous_admission",
            "readmission_30_days"
        ]
    ].head(20)
)
```

#  HOSPITAL KPI EVALUATION

```python
total_patients = df["patient_id"].nunique()

total_admissions = df["admission_id"].nunique()

total_revenue = df["total_bill"].sum()

average_bill = df["total_bill"].mean()

average_los = df[
    "length_of_stay_numeric"
].mean()

readmission_rate = (
    (df["readmission_30_days"] == "Yes").mean()
    * 100
)

print("=" * 55)
print("HOSPITAL KPI EVALUATION")
print("=" * 55)

print(f"Total Unique Patients   : {total_patients:,}")
print(f"Total Admissions        : {total_admissions:,}")
print(f"Total Revenue           : ₹{total_revenue:,.2f}")
print(f"Average Patient Bill    : ₹{average_bill:,.2f}")
print(f"Average Length of Stay  : {average_los:.2f} days")
print(f"30-Day Readmission Rate : {readmission_rate:.2f}%")
```

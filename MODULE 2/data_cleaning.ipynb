```python
import pandas as pd
import numpy as np

df=pd.read_csv('../data/hospital_raw_data.csv')
df.head()
```

```python
df=df.rename(columns={'DOB':'Birth_Year'})
df['Age'] = 2026 - df['Birth_Year']
df['Gender'] = df['Gender'].replace({'M':'Male','F':'Female'})
df['InsurancePlanName'] = 'Standard'
df['Payer_Mix'] = np.where(df['Insurance_Payer'] == 'Self-Pay', 'Self-Pay', 'Insurance')
df['Surgical_Mix'] = 'Medical' 
df['Discharge_Time'] = 'Not Applicable'
df['Discharge_Before_12PM'] = 'Not Applicable'
df.head()
```

```python
print("\n--- Data Quality Check ---")
total_records = len(df)
duplicate_count = df.duplicated(subset=['Case_No']).sum()
missing_values_count = df.isnull().any(axis=1).sum()
duplicate_error_pct = (duplicate_count / total_records) * 100
missing_error_pct = (missing_values_count / total_records) * 100
print(f"Total Records Before Cleaning: {total_records}")
print(f"Duplicate Records Found: {duplicate_count} ({duplicate_error_pct:.2f}%)")
print(f"Rows with Missing Values: {missing_values_count} ({missing_error_pct:.2f}%)")
print("--------------------------\n")
```

```python
df = df.drop_duplicates(subset=['Case_No'], keep='first')
df['Duplicate_Case_Flag'] = 'No'
total_records_after_dedup = len(df)
print(f"Total Records After Deduplication: {total_records_after_dedup}")
```

```python
final_columns = [
    'Month', 'Case_No', 'Birth_Year', 'Age', 'Nationality', 'Gender', 
    'DoctorLicense', 'DoctorName', 'Doctor_Type', 'Doctor_Status', 
    'CMI_Value', 'Specialty', 'Insurance_Payer', 'InsurancePlanName', 
    'Payer_Mix', 'Case_Type', 'LOS', 'Severity', 'Surgical_Mix', 
    'Revenue'
]

df_final = df[final_columns]

print("Saving cleaned dataset...")
output_path = '../data/hospital_data_AFTER_cleaning.csv'
df_final.to_csv(output_path, index=False)
print(f"Cleaned dataset saved to {output_path}")
```

Import libraries

```python
import pandas as pd
import numpy as np
from pathlib import Path
```

Load Dataset


```python
from pathlib import Path

BASE_DIR = Path.cwd()

INPUT_FILE = BASE_DIR / "hospital_raw_data.csv"

df = pd.read_csv(INPUT_FILE)

print("Dataset loaded successfully")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.head()
```

```python
df.info()
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

Check duplicate records

```python
duplicate_count = df.duplicated().sum()

print("Duplicate records:", duplicate_count)
```

Remove duplicate records

```python
df = df.drop_duplicates()

print("Rows after removing duplicates:", len(df))
```

Check missing values

```python
missing_count = df.isnull().sum()

missing_percentage = (
    df.isnull().mean() * 100
).round(2)

missing_report = pd.DataFrame({
    "Missing_Count": missing_count,
    "Missing_Percentage": missing_percentage
})

missing_report.sort_values(
    "Missing_Percentage",
    ascending=False
)
```

Standardize column name

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)

df.columns.tolist()
```

Standardize text columns

```python
text_columns = df.select_dtypes(
    include=["object"]
).columns

for column in text_columns:
    df[column] = (
        df[column]
        .astype("string")
        .str.strip()
    )

print("Text values standardized.")
```

Standardize department names

```python
if "department_name" in df.columns:

    df["department_name"] = (
        df["department_name"]
        .str.strip()
        .str.title()
    )

    print(
        df["department_name"]
        .value_counts()
    )
```

Standardize gender

```python
if "gender" in df.columns:

    df["gender"] = (
        df["gender"]
        .str.strip()
        .str.title()
    )

    gender_mapping = {
        "M": "Male",
        "MALE": "Male",
        "F": "Female",
        "FEMALE": "Female",
        "O": "Other",
        "OTHER": "Other"
    }

    df["gender"] = df["gender"].replace(
        gender_mapping
    )

df["gender"].value_counts(dropna=False)
```

Convert date columns

```python
date_columns = [
    "admission_date",
    "discharge_date",
    "date_of_birth"
]

for column in date_columns:

    if column in df.columns:

        df[column] = pd.to_datetime(
            df[column],
            errors="coerce"
        )

df[date_columns].dtypes
```

Check invalid dates

```python
if (
    "admission_date" in df.columns
    and "discharge_date" in df.columns
):

    invalid_dates = (
        df["discharge_date"]
        < df["admission_date"]
    )

    print(
        "Invalid discharge dates:",
        invalid_dates.sum()
    )
```

Correct invalid dates

```python
if (
    "admission_date" in df.columns
    and "discharge_date" in df.columns
):

    invalid_dates = (
        df["discharge_date"]
        < df["admission_date"]
    )

    df.loc[
        invalid_dates,
        "discharge_date"
    ] = pd.NaT
```

Calculate Length of Stay

```python
if (
    "admission_date" in df.columns
    and "discharge_date" in df.columns
):

    df["length_of_stay_days"] = (
        df["discharge_date"]
        - df["admission_date"]
    ).dt.days

df["length_of_stay_days"].describe()
```

Handle invalid Length of Stay

```python
if "length_of_stay_days" in df.columns:

    df.loc[
        df["length_of_stay_days"] < 0,
        "length_of_stay_days"
    ] = np.nan

print(
    "Invalid LOS values:",
    df["length_of_stay_days"].isna().sum()
)
```

Calculate patient age

```python
if (
    "admission_date" in df.columns
    and "date_of_birth" in df.columns
):

    df["patient_age"] = (
        (
            df["admission_date"]
            - df["date_of_birth"]
        ).dt.days / 365.25
    ).round(0)

df["patient_age"].describe()
```

Validate patient age

```python
if "patient_age" in df.columns:

    invalid_age = (
        (df["patient_age"] < 0)
        | (df["patient_age"] > 120)
    )

    print(
        "Invalid age values:",
        invalid_age.sum()
    )

    df.loc[
        invalid_age,
        "patient_age"
    ] = np.nan
```

Create age groups

```python
if "patient_age" in df.columns:

    df["age_group"] = pd.cut(
        df["patient_age"],
        bins=[0, 17, 30, 45, 60, 75, 120],
        labels=[
            "0-17",
            "18-30",
            "31-45",
            "46-60",
            "61-75",
            "76+"
        ],
        include_lowest=True
    )

df["age_group"].value_counts()
```

Create time variables

```python
if "admission_date" in df.columns:

    df["admission_year"] = (
        df["admission_date"].dt.year
    )

    df["admission_month"] = (
        df["admission_date"].dt.month
    )

    df["admission_month_name"] = (
        df["admission_date"].dt.month_name()
    )

    df["admission_quarter"] = (
        "Q"
        + df["admission_date"]
        .dt.quarter
        .astype("string")
    )

df[
    [
        "admission_year",
        "admission_month",
        "admission_month_name",
        "admission_quarter"
    ]
].head()
```

Handle missing categorical values

```python
categorical_columns = df.select_dtypes(
    include=["object", "string", "category"]
).columns

for column in categorical_columns:

    if df[column].isna().sum() > 0:

        df[column] = (
            df[column]
            .astype("string")
            .fillna("Unknown")
        )

print("Categorical missing values handled.")
```

Handle missing numeric values

```python
numeric_columns = df.select_dtypes(
    include=["number"]
).columns

for column in numeric_columns:

    if df[column].isna().sum() > 0:

        median_value = df[column].median()

        df[column] = df[column].fillna(
            median_value
        )

print("Numeric missing values handled.")
```

Check missing values again

```python
final_missing = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (
        df.isnull().mean() * 100
    ).round(2)
})

final_missing.sort_values(
    "Missing_Percentage",
    ascending=False
)
```

Check the required <2% condition

```python
total_cells = df.shape[0] * df.shape[1]

total_missing = df.isnull().sum().sum()

overall_missing_percentage = (
    total_missing / total_cells
) * 100

print(
    f"Overall missing percentage: "
    f"{overall_missing_percentage:.2f}%"
)

if overall_missing_percentage < 2:
    print("PASS: Missing values are below 2%.")
else:
    print("REVIEW: Missing values are 2% or higher.")
```

Final duplicate check

```python
print(
    "Remaining duplicate rows:",
    df.duplicated().sum()
)
```

Final dataset inspection

```python
print("Final rows:", df.shape[0])
print("Final columns:", df.shape[1])

df.head()
```

Save

```python
OUTPUT_FILE = BASE_DIR / "hospital_cleaned.csv"

df.to_csv(
    OUTPUT_FILE,
    index=False
)

print(
    "Cleaned dataset saved successfully:"
)

print(OUTPUT_FILE)
```

```python
#DATA TRANSFORMATION

data["DOB"] = pd.to_numeric(data["DOB"], errors="coerce")

data["DOB"] = pd.to_datetime(
    data["DOB"].astype("Int64").astype(str),
    format="%Y",
    errors="coerce"
)
```

```python
data["DOB"].head(10)
```

```python
current_year = pd.Timestamp.today().year

data["Age"] = current_year - data["DOB"].dt.year
```

```python
data[["DOB", "Age"]].head(10)
```

```python
bins = [0, 18, 30, 45, 60, 75, 120]

labels = ["0-18", "19-30", "31-45", "46-60", "61-75", "76+"]

data["Age Group"] = pd.cut(
    data["Age"],
    bins=bins,
    labels=labels,
    right=True
)
```

```python
# Hospital KPIs

total_cases = df["Case_No"].count()
total_revenue = df["Revenue"].sum()
average_revenue = df["Revenue"].mean()
average_los = df["LOS"].mean()
average_cmi = df["CMI Value"].mean()

print("===== HOSPITAL KPIs =====")
print("Total Cases       :", total_cases)
print("Total Revenue     :", round(total_revenue, 2))
print("Average Revenue   :", round(average_revenue, 2))
print("Average LOS       :", round(average_los, 2))
print("Average CMI Value :", round(average_cmi, 2))
```

```python
kpis = pd.DataFrame({
    "KPI": [
        "Total Cases",
        "Total Revenue",
        "Average Revenue",
        "Average LOS",
        "Average CMI Value"
    ],
    "Value": [
        total_cases,
        total_revenue,
        average_revenue,
        average_los,
        average_cmi
    ]
})

kpis
```

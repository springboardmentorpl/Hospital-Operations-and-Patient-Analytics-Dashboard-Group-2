```python
import pandas as pd

df = pd.read_csv("hospital_cleaned_dataset.csv")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
print(df.columns.tolist())
```

```python
total_admissions = len(df)

print("Total Admissions =", total_admissions)
```

```python
occupancy_rate = df["Occupancy_Rate"].mean()

print("Occupancy Rate =", occupancy_rate)
```

```python
avg_los = df["Length_of_Stay"].mean()

print("Average Length of Stay =", avg_los)
```

```python
readmission_rate = (
    df["Readmission_Within_30_Days"].sum()
    / len(df)
) * 100

print("Readmission Rate =", readmission_rate)
```

```python
print(df["Readmission_Within_30_Days"].unique())
```

```python
readmission_rate = (
    df["Readmission_Within_30_Days"]
      .astype(str)
      .str.strip()
      .str.upper()
      .eq("YES")
      .sum()
    / len(df)
) * 100

print("Readmission Rate =", round(readmission_rate, 2), "%")
```

```python
bed_utilization_rate = (
    df["Beds_Occupied"].sum()
    / df["Total_Beds_Available"].sum()
) * 100

print("Bed Utilization Rate =", round(bed_utilization_rate, 2), "%")
```

```python
department_efficiency = (
    df.groupby("Department")["Length_of_Stay"]
      .mean()
      .reset_index()
)

department_efficiency.columns = [
    "Department",
    "Average_Length_of_Stay"
]

department_efficiency
```

```python
bed_utilization_rate = (
    df["Length_of_Stay"].sum()
    /
    (df["Available_Beds"].sum() * 30)
) * 100

print("Bed Utilization Rate =", round(bed_utilization_rate,2))
```

```python
print(df.columns.tolist())
```

```python
bed_utilization_rate = (
    df["Beds_Occupied"].sum()
    /
    df["Total_Beds_Available"].sum()
) * 100

print("Bed Utilization Rate =", round(bed_utilization_rate,2))
```

```python
kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Admissions",
        "Occupancy Rate",
        "Average Length of Stay",
        "Readmission Rate",
        "Bed Utilization Rate"
    ],
    "Value": [
        total_admissions,
        occupancy_rate,
        avg_los,
        readmission_rate,
        bed_utilization_rate
    ]
})

kpi_summary
```

```python
with pd.ExcelWriter("hospital_final_dataset.xlsx") as writer:

    df.to_excel(
        writer,
        sheet_name="Cleaned_Data",
        index=False
    )

    kpi_summary.to_excel(
        writer,
        sheet_name="KPI_Summary",
        index=False
    )

    department_efficiency.to_excel(
        writer,
        sheet_name="Department_Efficiency",
        index=False
    )

print("Excel File Saved Successfully")
```

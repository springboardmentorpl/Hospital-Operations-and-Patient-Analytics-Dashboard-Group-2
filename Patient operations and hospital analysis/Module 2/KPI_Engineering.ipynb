```python
import pandas as pd
import numpy as np

# Load hospital dataset
df = pd.read_csv("Hospital_Operations_Final_Dataset.csv")
df.head()
```

```python
# Calculate Length of Stay in days
df["length_of_stay"] = (
    df["departure_date"] - df["arrival_date"]
).dt.days

print(df["length_of_stay"].describe())
```

```python
total_admissions = df["patient_id"].nunique()

print("Total Admissions:", total_admissions)
```

```python
average_length_of_stay = df["length_of_stay"].mean()

print(
    "Average Length of Stay:",
    round(average_length_of_stay, 2),
    "days"
)
```

```python
total_admitted = df["total_patients_admitted"].sum()
total_available_beds = df["total_available_beds"].sum()

bed_utilization_rate = (
    total_admitted / total_available_beds
) * 100

print(
    "Bed Utilization Rate:",
    round(bed_utilization_rate, 2),
    "%"
)
```

```python
service_kpi = (
    df[
        [
            "service",
            "total_available_beds",
            "total_patients_admitted"
        ]
    ]
    .drop_duplicates()
)

service_kpi["bed_utilization_rate"] = (
    service_kpi["total_patients_admitted"]
    / service_kpi["total_available_beds"]
) * 100

print(service_kpi)
```

```python
occupancy_rate = (
    service_kpi["total_patients_admitted"].sum()
    / service_kpi["total_available_beds"].sum()
) * 100

print("Occupancy Rate:", round(occupancy_rate, 2), "%")
```

```python
patient_counts = df["patient_id"].value_counts()

readmitted_patients = patient_counts[patient_counts > 1].count()

total_patients = df["patient_id"].nunique()

readmission_rate = (
    readmitted_patients / total_patients
) * 100

print("Readmitted Patients:", readmitted_patients)
print("Readmission Rate:", round(readmission_rate, 2), "%")
```

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)
```

```python
# LOAD DATASETS
hospital_patient = pd.read_csv(
    "Hospital Patient .csv"
)

patients = pd.read_csv(
    "patients.csv"
)

services_weekly = pd.read_csv(
    "services_weekly.csv"
)

staff = pd.read_csv(
    "staff.csv"
)

staff_schedule = pd.read_csv(
    "staff_schedule.csv"
)

print("All datasets loaded successfully!")
```

```python
# DATASET SHAPE BEFORE CLEANING

datasets = {
    "Hospital Patient": hospital_patient,
    "Patients": patients,
    "Services Weekly": services_weekly,
    "Staff": staff,
    "Staff Schedule": staff_schedule
}

print("DATASET INFORMATION BEFORE CLEANING")


for name, df in datasets.items():

    print(name)

    print("Rows    :", df.shape[0])
    print("Columns :", df.shape[1])
```

```python
# COLUMN INFORMATION
for name, df in datasets.items():
    print(name)
    for i, column in enumerate(df.columns, 1):
        print(f"{i}. {column}")
```

```python
# NULL VALUE CHECK

print("=" * 70)
print("NULL VALUE ANALYSIS - BEFORE CLEANING")
print("=" * 70)

for name, df in datasets.items():

    print("\n" + "-" * 60)
    print(name)
    print("-" * 60)

    null_count = df.isnull().sum()

    null_report = pd.DataFrame({
        "Column": df.columns,
        "Null_Count": null_count.values,
        "Null_Percentage": (
            null_count.values / len(df) * 100
        ).round(2)
    })

    display(null_report)
```

```python
# DUPLICATE ROW CHECK

print("=" * 70)
print("DUPLICATE CHECK - BEFORE CLEANING")
print("=" * 70)

for name, df in datasets.items():

    duplicate_count = df.duplicated().sum()

    print(
        f"{name:<25} → {duplicate_count} duplicate rows"
    )
```

```python
# UNIQUE ID CHECK

print("=" * 70)
print("ID UNIQUENESS CHECK")
print("=" * 70)

print("\nHospital Patient")
print("Total IDs  :", hospital_patient["Patient_ID"].count())
print("Unique IDs :", hospital_patient["Patient_ID"].nunique())

print("\nPatients")
print("Total IDs  :", patients["patient_id"].count())
print("Unique IDs :", patients["patient_id"].nunique())

print("\nStaff")
print("Total IDs  :", staff["staff_id"].count())
print("Unique IDs :", staff["staff_id"].nunique())
```

```python
# CHECK COMMON SERVICE VALUES

print("=" * 70)
print("SERVICE VALUE CHECK")
print("=" * 70)

print("\nPatients:")
print(patients["service"].unique())

print("\nServices Weekly:")
print(services_weekly["service"].unique())

print("\nStaff:")
print(staff["service"].unique())
```

```python
# STANDARDIZE COLUMN NAMES

def clean_column_names(df):

    df = df.copy()

    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
        .str.replace("-", "_")
    )

    return df


hospital_patient = clean_column_names(hospital_patient)
patients = clean_column_names(patients)
services_weekly = clean_column_names(services_weekly)
staff = clean_column_names(staff)
staff_schedule = clean_column_names(staff_schedule)

print("Column names standardized!")
```

```python
# DATE CONVERSION

hospital_patient["admission_date"] = pd.to_datetime(
    hospital_patient["admission_date"],
    dayfirst=True,
    errors="coerce"
)

hospital_patient["discharge_date"] = pd.to_datetime(
    hospital_patient["discharge_date"],
    dayfirst=True,
    errors="coerce"
)

patients["arrival_date"] = pd.to_datetime(
    patients["arrival_date"],
    errors="coerce"
)

patients["departure_date"] = pd.to_datetime(
    patients["departure_date"],
    errors="coerce"
)

print("Date conversion completed!")
```

```python
# CHECK INVALID DATES

print("=" * 70)
print("DATE VALIDATION")
print("=" * 70)

print("\nHospital Patient:")
print(
    hospital_patient[
        ["admission_date", "discharge_date"]
    ].isnull().sum()
)

print("\nPatients:")
print(
    patients[
        ["arrival_date", "departure_date"]
    ].isnull().sum()
)
```

```python
# SERVICE LEVEL AGGREGATION

service_summary = (
    services_weekly
    .groupby("service")
    .agg(
        total_available_beds=("available_beds", "sum"),
        total_patient_requests=("patients_request", "sum"),
        total_patients_admitted=("patients_admitted", "sum"),
        total_patients_refused=("patients_refused", "sum"),
        avg_patient_satisfaction=("patient_satisfaction", "mean"),
        avg_staff_morale=("staff_morale", "mean")
    )
    .reset_index()
)

print("=" * 70)
print("SERVICE SUMMARY")
print("=" * 70)

display(service_summary)
```

```python
# BED UTILIZATION

service_summary["bed_utilization_rate"] = (
    service_summary["total_patients_admitted"]
    /
    service_summary["total_available_beds"]
)

service_summary["bed_utilization_rate"] = (
    service_summary["bed_utilization_rate"]
    .clip(upper=1)
    .round(4)
)

display(service_summary)
```

```python
# STAFF SUMMARY

staff_summary = (
    staff
    .groupby("service")
    .agg(
        total_staff=("staff_id", "nunique"),
        doctors=("role", lambda x: (x == "doctor").sum()),
        nurses=("role", lambda x: (x == "nurse").sum())
    )
    .reset_index()
)

print("=" * 70)
print("STAFF SUMMARY")
print("=" * 70)

display(staff_summary)
```

```python
# STAFF ATTENDANCE SUMMARY

attendance_summary = (
    staff_schedule
    .groupby("service")
    .agg(
        total_staff_schedule_records=("staff_id", "count"),
        total_present=("present", "sum"),
        attendance_rate=("present", "mean")
    )
    .reset_index()
)

attendance_summary["attendance_rate"] = (
    attendance_summary["attendance_rate"] * 100
).round(2)

display(attendance_summary)
```

```python
# COMBINE SERVICE TABLES

service_operations = service_summary.merge(
    staff_summary,
    on="service",
    how="left"
)

service_operations = service_operations.merge(
    attendance_summary,
    on="service",
    how="left"
)

print("=" * 70)
print("COMBINED SERVICE OPERATIONS TABLE")
print("=" * 70)

print("Shape :", service_operations.shape)

display(service_operations)
```

```python
# CREATE FINAL PATIENT OPERATIONS DATASET

final_dataset = patients.merge(
    service_operations,
    on="service",
    how="left"
)

print("=" * 70)
print("FINAL COMBINED DATASET")
print("=" * 70)

print("Rows    :", final_dataset.shape[0])
print("Columns :", final_dataset.shape[1])

display(final_dataset.head())
```

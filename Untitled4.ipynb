```python
import pandas as pd
import glob

files = glob.glob("/content/*.csv")

print("CSV files found:")
print(files)
```

```python
hospital_file = [f for f in files if "Hospital" in f][0]

df = pd.read_csv(hospital_file)

print("✅ Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())
```

```python
df.head
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
missing_percent = (df.isnull().sum() / len(df)) * 100
print(missing_percent)
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
df=df.drop_duplicates()
```

```python
print("Duplicates after removal:", df.duplicated().sum())
```

```python
df.columns = df.columns.str.strip()
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ", "_")
```

```python
print(df.columns.tolist())
```

```python
for i, col in enumerate(df.columns, 1):
    print(i, col)
```

```python
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()
```

```python
print([col for col in df.columns if "gender" in col or "sex" in col])
```

```python
import matplotlib.pyplot as plt

df['gender'].value_counts().plot(kind='bar')

plt.title("Male vs Female Patients")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")
plt.show()
```

```python
print("Final dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
missing_percent = (df.isnull().sum() / len(df)) * 100

print(missing_percent)
```

```python
threshold = 2

missing_percent = (df.isnull().sum() / len(df)) * 100

columns_to_remove = missing_percent[missing_percent > threshold].index

df = df.drop(columns=columns_to_remove)

print("Removed columns:")
print(columns_to_remove.tolist())
```

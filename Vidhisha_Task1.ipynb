```python
import pandas as pd

df = pd.read_csv('hospital_operations_dataset.csv')

df.head()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
# Remove duplicates
df = df.drop_duplicates()

# Fill missing values
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])

df['Department'] = df['Department'].fillna(df['Department'].mode()[0])

df['Length_of_Stay'] = df['Length_of_Stay'].fillna(df['Length_of_Stay'].median())

df['Oxygen_Units_Used'] = df['Oxygen_Units_Used'].fillna(df['Oxygen_Units_Used'].median())

# Standardize text columns
text_columns = ['Gender','Severity','Admission_Type','Department','Day_of_Week']

for col in text_columns:
    df[col] = df[col].astype(str).str.strip().str.title()

print("Cleaning Completed Successfully!")
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
# Create Age Group

df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[0,18,40,60,100],
    labels=['Child','Young Adult','Adult','Senior']
)

# Create Bed Occupancy Rate

df['Occupancy_Rate'] = (
    df['Beds_Occupied'] /
    df['Total_Beds_Available']
) * 100

df[['Age','Age_Group','Occupancy_Rate']].head()
```

```python
print("Total Patients:", len(df))
print("Average Length of Stay:", round(df['Length_of_Stay'].mean(),2))
print("Average Oxygen Usage:", round(df['Oxygen_Units_Used'].mean(),2))
print("Total Departments:", df['Department'].nunique())
print("Total Beds Occupied:", df['Beds_Occupied'].sum())
print("Total ICU Beds Occupied:", df['ICU_Beds_Occupied'].sum())
```

```python
df.to_csv("hospital_cleaned_dataset.csv", index=False)
```

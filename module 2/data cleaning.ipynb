```python
# Step 1 - Import Required Libraries
import pandas as pd
import numpy as np
print("Libraries imported successfully.")
```

```python
# Load directly using the file name
dataset_name = "Hospital_RawDataset_Updated.csv"
hospital_df = pd.read_csv(dataset_name)

print(f"Dataset '{dataset_name}' loaded successfully.")
```

```python
# Get the number of rows and columns
rows, columns = hospital_df.shape

print("Number of Records (Rows):", rows)
print("Number of Attributes (Columns):", columns)
```

```python
# Display the first 5 records
print("First 5 Records:")
display(hospital_df.head())
```

```python
# Display the last 5 records
print("Last 5 Records:")
display(hospital_df.tail())
```

```python
# Display all column names
print("Dataset Columns:\n")

for index, column in enumerate(hospital_df.columns, start=1):
    print(f"{index}. {column}")
```

```python
# Display data types and non-null counts
hospital_df.info()
```

```python
# STEP 2 - MISSING VALUE ANALYSIS
# Calculate the number of missing values in each column
missing_count = hospital_df.isnull().sum()

# Calculate missing percentage for each column
missing_percentage = (missing_count / len(hospital_df)) * 100

# Create a summary table
missing_summary = pd.DataFrame({
    "Missing Values": missing_count,
    "Missing Percentage": missing_percentage
})

# Display only columns containing missing values
missing_summary = missing_summary[
    missing_summary["Missing Values"] > 0
].sort_values(
    by="Missing Percentage",
    ascending=False
)

print("Missing Value Summary:")
display(missing_summary)
```

```python
# DUPLICATE RECORD ANALYSIS
# Count completely duplicated rows
duplicate_count = hospital_df.duplicated().sum()

print("Number of Duplicate Records:", duplicate_count)

# Calculate duplicate percentage
duplicate_percentage = (
    duplicate_count / len(hospital_df)
) * 100

print(
    f"Duplicate Percentage: {duplicate_percentage:.2f}%"
)
```

```python
# INITIAL DATA QUALITY SUMMARY
total_cells = hospital_df.shape[0] * hospital_df.shape[1]
total_missing = hospital_df.isnull().sum().sum()

# Calculate overall completeness
completeness_percentage = (
    (total_cells - total_missing) / total_cells
) * 100

print("==============================================")
print("   INITIAL HOSPITAL DATA QUALITY SUMMARY")
print("==============================================")
print(f"Total Records       : {hospital_df.shape[0]}")
print(f"Total Attributes    : {hospital_df.shape[1]}")
print(f"Total Missing Cells : {total_missing}")
print(f"Duplicate Records   : {duplicate_count}")
print(f"Data Completeness   : {completeness_percentage:.2f}%")
print("==============================================")
```

```python
# STEP 6. - HANDLE MISSING TRANSFER DEPARTMENT
# Create a copy of the dataset if not already done
clean_df = hospital_df.copy()

# Handle missing Transfer_To_Department values
clean_df["Transfer_To_Department"] = clean_df["Transfer_To_Department"].fillna("Not Transferred")

print("Missing Transfer_To_Department values handled.")
print("Remaining missing values:", clean_df["Transfer_To_Department"].isnull().sum())
```

```python
# STEP 3 - HANDLE MISSING TRANSFER DATE
# A missing transfer date means that no transfer occurred.
# We preserve this information using a meaningful categorical label.

clean_df["Transfer_Date"] = (
    clean_df["Transfer_Date"]
    .fillna("Not Transferred")
)

print("Missing Transfer_Date values handled.")
print(
    "Remaining missing values:",
    clean_df["Transfer_Date"].isnull().sum()
)
```

```python
# HANDLE MISSING INSURANCE INFORMATION
# Missing insurance information is retained as an explicit
# category rather than deleting patient records.

clean_df["Insurance Provider"] = (
    clean_df["Insurance Provider"]
    .fillna("Not Available")
)

print("Missing Insurance Provider values handled.")
print(
    "Remaining missing values:",
    clean_df["Insurance Provider"].isnull().sum()
)
```

```python
# STEP 6.4 - VERIFY MISSING VALUE TREATMENT
# Calculate remaining missing values
remaining_missing = clean_df.isnull().sum()

# Keep only columns that still contain missing values
remaining_missing = remaining_missing[
    remaining_missing > 0
].sort_values(ascending=False)


print("POST-TREATMENT MISSING VALUE ANALYSIS")


if len(remaining_missing) == 0:
    print("No missing values remain in the dataset.")
else:
    print("Columns with remaining missing values:")
    display(remaining_missing)

```

```python
# CALCULATE POST-TREATMENT COMPLETENESS

total_cells_after = (
    clean_df.shape[0] * clean_df.shape[1]
)

missing_cells_after = (
    clean_df.isnull().sum().sum()
)

completeness_after = (
    (total_cells_after - missing_cells_after)
    / total_cells_after
) * 100

missing_percentage_after = (
    missing_cells_after / total_cells_after
) * 100

print("=" * 60)
print("       DATA COMPLETENESS AFTER MISSING VALUE TREATMENT")
print("=" * 60)

print(f"Total Cells          : {total_cells_after}")
print(f"Missing Cells        : {missing_cells_after}")
print(f"Missing Percentage   : {missing_percentage_after:.2f}%")
print(f"Data Completeness    : {completeness_after:.2f}%")

print("=" * 60)
```

```python
#  INSPECT CURRENT DEPARTMENT NAMES
# Display all unique department names before standardization.
# This allows us to identify spelling, spacing, or capitalization
# inconsistencies before modifying the data.

department_before = (
    clean_df["Department"]
    .value_counts(dropna=False)
    .reset_index()
)

department_before.columns = [
    "Department",
    "Record_Count"
]
print(" DEPARTMENT NAMES BEFORE STANDARDIZATION")
display(department_before)
print("Total Unique Department Values:",
      clean_df["Department"].nunique())
```

```python
# CHECK FOR UNNECESSARY SPACES
# Identify department values containing leading or trailing spaces.
space_issues = clean_df[
    clean_df["Department"] != clean_df["Department"].str.strip()
]
print("DEPARTMENT SPACE VALIDATION")
print(
    "Records with leading/trailing spaces:",
    len(space_issues)
)
```

```python
# STANDARDIZE DEPARTMENT TEXT FORMAT
# Remove leading and trailing whitespace.
clean_df["Department"] = (
    clean_df["Department"]
    .astype(str)
    .str.strip()
)

# Replace multiple consecutive spaces with a single space.
clean_df["Department"] = (
    clean_df["Department"]
    .str.replace(r"\s+", " ", regex=True)
)

# Standardize capitalization using title case.
clean_df["Department"] = (
    clean_df["Department"]
    .str.title()
)
print("Department text formatting standardized successfully.")
```

```python
# Convert date columns into datetime format
clean_df["Admission Date"] = pd.to_datetime(
    clean_df["Admission Date"], errors="coerce"
)

clean_df["Discharge Date"] = pd.to_datetime(
    clean_df["Discharge Date"], errors="coerce"
)

# Transfer_Date contains "Not Transferred", so convert only
# actual date values and preserve the meaningful text label.
clean_df["Transfer_Date"] = pd.to_datetime(
    clean_df["Transfer_Date"],
    errors="coerce"
)

print("Date standardization completed successfully. ")
```

```python
# Check the resulting data types
print(clean_df[
    ["Admission Date", "Discharge Date", "Transfer_Date"]
].dtypes)
```

## Data Transformation

Create derived healthcare attributes to make the dataset suitable
for analysis and dashboard development.

```python
# Create Age Group
clean_df["Age Group"] = pd.cut(
    clean_df["Age"],
    bins=[0, 18, 35, 50, 65, 120],
    labels=[
        "0-18",
        "19-35",
        "36-50",
        "51-65",
        "65+"
    ]
)

# Create Admission Month
clean_df["Admission Month"] = (
    clean_df["Admission Date"].dt.month_name()
)

# Create Admission Year
clean_df["Admission Year"] = (
    clean_df["Admission Date"].dt.year
)

# Create Length of Stay Category
clean_df["LOS Category"] = pd.cut(
    clean_df["Length of Stay"],
    bins=[-1, 3, 7, 14, float("inf")],
    labels=[
        "Short Stay",
        "Medium Stay",
        "Long Stay",
        "Extended Stay"
    ]
)

print("Data transformation completed successfully. ")
```

```python
# Display the newly created columns
clean_df[
    [
        "Age",
        "Age Group",
        "Admission Month",
        "Admission Year",
        "Length of Stay",
        "LOS Category"
    ]
].head()
```

```python
# FIX - RESTORE NON-TRANSFERRED STATUS
# Patients without a transfer should have a meaningful label
# instead of a missing date value.

clean_df.loc[
    clean_df["Transfer_To_Department"] == "Not Transferred",
    "Transfer_Date"
] = "Not Transferred"

print("Transfer date values corrected successfully. ✅")
print("Remaining missing values:", clean_df.isnull().sum().sum())
```

```python
# STEP 12 - FINAL DATA QUALITY VALIDATION
final_missing = clean_df.isnull().sum().sum()
final_duplicates = clean_df.duplicated().sum()


print("FINAL DATA QUALITY CHECK")


print("Records          :", len(clean_df))
print("Columns          :", len(clean_df.columns))
print("Missing Values   :", final_missing)
print("Duplicates       :", final_duplicates)

if final_missing == 0 and final_duplicates == 0:
    print("\nData Quality Check: PASSED ✅")
else:
    print("\nData Quality Check: REVIEW REQUIRED ⚠️")
```

```python
# STEP 13 - EXPORT CLEANED DATASET
import os

# Create output folder if it doesn't exist
output_dir = "output"
os.makedirs(output_dir, exist_ok=True)

# Export cleaned dataframe to CSV
cleaned_file_path = os.path.join(output_dir, "hospital_cleaned.csv")
clean_df.to_csv(cleaned_file_path, index=False)


print("EXPORT SUCCESSFUL 🚀")

print(f"File Saved At: {cleaned_file_path}")
print(f"Total Rows   : {clean_df.shape[0]}")
print(f"Total Columns: {clean_df.shape[1]}")
```

```python
# STEP 14 - VERIFY EXPORTED CSV
# Read back exported CSV
exported_df = pd.read_csv(cleaned_file_path)

print("Verified Rows   :", len(exported_df))
print("Verified Columns:", len(exported_df.columns))
print("Null Values     :", exported_df.isnull().sum().sum())

if len(exported_df) == len(clean_df) and exported_df.isnull().sum().sum() == 0:
    print("\n MILESTONE 1 COMPLETE! Deliverables are ready.")
else:
    print("\n⚠️ Verification failed. Please re-check the export step.")
```

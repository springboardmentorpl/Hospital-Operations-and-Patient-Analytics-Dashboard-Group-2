```python
import pandas as pd

df = pd.read_csv("hospital_cleaned.csv")

print(df.shape)
print(df.columns.tolist())
```

```python
print(df.head())
print(df.info())
```

```python

# ============================================================
# 3. HOSPITAL OVERVIEW KPIs
# ============================================================

total_admissions = len(df)

total_revenue = df["Billing Amount"].sum()

average_los = df["Length of Stay"].mean()

median_los = df["Length of Stay"].median()

max_los = df["Length of Stay"].max()

average_age = df["Age"].mean()

readmitted_patients = df["Readmission"].eq("Yes").sum()

readmission_rate = (
    readmitted_patients / total_admissions
) * 100

occupied_records = df["Bed Occupied"].eq("Yes").sum()

bed_utilization = (
    occupied_records / total_admissions
) * 100

staff_utilization = (
    df["Staff_Utilization_%_Derived"].mean()
)

bed_occupancy = (
    df["Bed_Occupancy_Rate_%"].mean() * 100
)


print("\nHOSPITAL OVERVIEW KPIs")
print("=" * 60)

print("Total Admissions:", total_admissions)
print("Unique Patients:", df["Patient ID"].nunique())
print("Total Revenue:", total_revenue)
print("Average LOS:", average_los)
print("Median LOS:", median_los)
print("Maximum LOS:", max_los)
print("Average Age:", average_age)
print("Readmitted Patients:", readmitted_patients)
print("Readmission Rate:", readmission_rate, "%")
print("Bed Utilization:", bed_utilization, "%")
print("Average Staff Utilization:", staff_utilization, "%")
print("Average Bed Occupancy:", bed_occupancy, "%")


# ============================================================
# 4. RESOURCE UTILIZATION
# ============================================================

print("\nRESOURCE UTILIZATION")
print("=" * 60)

print("Average Beds Available:", df["Beds Available"].mean())
print("Total Beds Available (sum):", df["Beds Available"].sum())

print("Average Department Bed Capacity:",
      df["Dept_Bed_Capacity_Derived"].mean())

print("Average Beds Occupied:",
      df["Beds_Occupied_Count"].mean())

print("Average ICU Capacity:",
      df["Dept_ICU_Bed_Capacity_Derived"].mean())

print("Average Staff Capacity:",
      df["Dept_Staff_Capacity_Derived"].mean())

print("Average Staff Utilization:",
      df["Staff_Utilization_%_Derived"].mean())

print("Average Bed Occupancy:",
      df["Bed_Occupancy_Rate_%"].mean() * 100)


# ============================================================
# 5. DEPARTMENT ANALYTICS
# ============================================================

print("\nPATIENTS BY DEPARTMENT")
print("=" * 60)

patients_by_department = (
    df.groupby("Department")["Patient ID"]
      .count()
      .sort_values(ascending=False)
)

print(patients_by_department)


print("\nAVERAGE LOS BY DEPARTMENT")
print("=" * 60)

avg_los_department = (
    df.groupby("Department")["Length of Stay"]
      .mean()
      .sort_values(ascending=False)
)

print(avg_los_department)


print("\nSTAFF UTILIZATION BY DEPARTMENT")
print("=" * 60)

staff_department = (
    df.groupby("Department")["Staff_Utilization_%_Derived"]
      .mean()
      .sort_values(ascending=False)
)

print(staff_department)


print("\nBED OCCUPANCY BY DEPARTMENT")
print("=" * 60)

bed_department = (
    df.groupby("Department")["Bed_Occupancy_Rate_%"]
      .mean()
      .sort_values(ascending=False)
)

print(bed_department * 100)


# ============================================================
# 6. PATIENT FLOW
# ============================================================

df["Admission Date"] = pd.to_datetime(
    df["Admission Date"],
    errors="coerce"
)

df["Discharge Date"] = pd.to_datetime(
    df["Discharge Date"],
    errors="coerce"
)

print("\nPATIENT FLOW")
print("=" * 60)

admissions_by_month = (
    df.groupby(
        df["Admission Date"].dt.to_period("M")
    )["Patient ID"]
    .count()
)

print("\nAdmissions by Month:")
print(admissions_by_month)


discharges_by_month = (
    df.groupby(
        df["Discharge Date"].dt.to_period("M")
    )["Patient ID"]
    .count()
)

print("\nDischarges by Month:")
print(discharges_by_month)


# ============================================================
# 7. LENGTH OF STAY DISTRIBUTION
# ============================================================

print("\nLENGTH OF STAY DISTRIBUTION")
print("=" * 60)

los_distribution = df["LOS Category"].value_counts()

print(los_distribution)


# ============================================================
# 8. ADMISSION TYPE
# ============================================================

print("\nADMISSION TYPE")
print("=" * 60)

print(df["Admission Type"].value_counts())


# ============================================================
# 9. STATE-WISE ADMISSIONS
# ============================================================

print("\nADMISSIONS BY STATE")
print("=" * 60)

state_admissions = (
    df.groupby("State")["Patient ID"]
      .count()
      .sort_values(ascending=False)
)

print(state_admissions)


# ============================================================
# 10. HOSPITAL-WISE ADMISSIONS
# ============================================================

print("\nADMISSIONS BY HOSPITAL")
print("=" * 60)

hospital_admissions = (
    df.groupby("Hospital Name")["Patient ID"]
      .count()
      .sort_values(ascending=False)
)

print(hospital_admissions)


# ============================================================
# 11. LOS INDEPENDENT VALIDATION
# ============================================================

df["LOS_Check"] = (
    df["Discharge Date"] - df["Admission Date"]
).dt.days

los_match = (
    df["Length of Stay"] == df["LOS_Check"]
).all()

print("\nLOS VALIDATION")
print("=" * 60)

print("Calculated LOS matches dataset:",
      los_match)

print("Average Calculated LOS:",
      df["LOS_Check"].mean())


# ============================================================
# 12. CREATE KPI VALIDATION TABLE
# ============================================================

validation = pd.DataFrame({
    "KPI": [
        "Total Admissions",
        "Unique Patients",
        "Total Revenue",
        "Average LOS",
        "Median LOS",
        "Maximum LOS",
        "Average Age",
        "Readmission Rate %",
        "Bed Utilization %",
        "Staff Utilization %",
        "Average Bed Occupancy %"
    ],

    "Python Value": [
        total_admissions,
        df["Patient ID"].nunique(),
        total_revenue,
        average_los,
        median_los,
        max_los,
        average_age,
        readmission_rate,
        bed_utilization,
        staff_utilization,
        bed_occupancy
    ]
})

print("\nFINAL KPI VALIDATION TABLE")
print("=" * 60)

print(validation.to_string(index=False))
```

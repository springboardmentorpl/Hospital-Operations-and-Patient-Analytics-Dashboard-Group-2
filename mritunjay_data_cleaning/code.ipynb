```python
df["Month"] = pd.to_datetime(df["Month"])
print(df["Month"].dtype)
print(df["Month"].head())
```

```python
df.columns
```

```python
print(df["Discharge Time"].dtype)
df["Discharge Time"] = pd.to_datetime(
    df["Discharge Time"],
    format="%H:%M:%S"
).dt.time

print(df["Discharge Time"].dtype)
print(df["Discharge Time"].head())
```

```python
print(df["DOB"].dtype)
df["DOB"].head()
```

```python
df["DOB"] = df["DOB"].astype(int)
print(df["DOB"].head())
```

```python
print(df.dtypes)
```

```python
print((df["Case_No"] % 1 == 0).all())
df["Case_No"] = df["Case_No"].astype(int)

print((df["Severity"] % 1 == 0).all())
df["Severity"] = df["Severity"].astype(int)

print(type(df["Discharge Time"].iloc[0]))
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)

print(df[["Case_No", "DOB", "CMI Value", "LOS", "Severity", "Revenue"]].describe())
```
